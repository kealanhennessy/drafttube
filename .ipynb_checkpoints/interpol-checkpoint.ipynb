```python
import numpy as np
from scipy.interpolate import interp2d
from reader import *
```

```python
geo = Reader("conetest.geo", cascade=False)
geo.collate_pt(0)
```

```python
pts = []
for i in range(geo.size()):
    pts.append(geo.collate_pt(i))
pts
```

```python
def unique(arr, seg=None):
    if seg is none:
        print("Unexpected error: must provide a segment")
        raise
    unique = []
    unique.append(arr[0][2])
    for i in range(len(arr)):
        if 2 <= len(unique):
            break
        if arr[i][2] in unique:
            continue
        else:
            unique.append(arr[i][2])
    return unique
```

```python
print(unique(pts))
```
