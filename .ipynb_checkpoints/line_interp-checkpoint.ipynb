```python
import numpy as np
import matplotlib.pyplot as plt
import math
from scipy import interpolate
from planeInterp import *
```

```python
baseline1 = planeInterpolator("1")
#allpoints = baseline1.cone_interp()
cone = baseline1.cone_interp() 
elbow = baseline1.elbow_interp()
connector = baseline1.cmd_interp("connector")
middle = baseline1.cmd_interp("middle")
diffuser = baseline1.cmd_interp("diffuser")
```

```python

k = 359
x = []
z = []
for i in range(len(cone)):
    if k<180:
        x.append(cone[i][k+180][0])
        z.append(cone[i][k+180][2])
    else:
        x.append(cone[i][k-180][0])
        z.append(cone[i][k-180][2])
for i in range(len(elbow)):
    x.append(elbow[i][k][0])
    z.append(elbow[i][k][2])
    if elbow[i][k][2] > 3:
        print(i)
for i in range(65):
    x.append(connector[i][k][0])
    z.append(connector[i][k][2])
    x.append(middle[i][k][0])
    z.append(middle[i][k][2])
    x.append(diffuser[i][k][0])
    z.append(diffuser[i][k][2])
plt.plot(x, z, '.')
#print(x)

```
