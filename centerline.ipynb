```python
shapes = ['baseline1','baseline4', 'baseline5', 'baseline6']
parts = ['cone', 'elbow', 'connector', 'middle', 'diffuser']

def get_points(reader):
    points = []
    for i in range(reader.size()):
        points.append(np.array(reader.collate_xyz(i)))
    points = np.asarray(points)
    return points

cone_geo = Reader("drafttube_shapes/" + "baseline1" + "/geo/" + "cone" + ".geo", cascade=True) 

def separate(pt_array, s, f):
    x = []
    y = []
    z = []
    for i in range(s, f):
        x.append(pt_array[i][0])
        y.append(pt_array[i][1])
        z.append(pt_array[i][2])
    return [np.mean(x), np.mean(y), np.mean(z)]

cone_geo = Reader("drafttube_shapes/" + "baseline1" + "/geo/" + "cone" + ".geo", cascade=True) 
pts = get_points(cone_geo)
print((cone_geo.size() - (cone_geo.size() % 256)))
#print(separate(pts, 0, 256))


#print(np.mean([pts[i][0] for i in range(256)]))
def cone(shape, part='cone'):
    cone_geo = Reader("drafttube_shapes/" + shape + "/geo/" + part + ".geo", cascade=True) 
    pts = get_points(cone_geo)
    means = []
    n = 0
    i = n + 256
    while i <= (cone_geo.size() - (cone_geo.size() % 256)):
        mean = separate(pts, n, i)
        means.append(mean)
        n += 256
    return means

print(cone('baseline1'))
```
