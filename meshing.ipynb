```python
k = 130
layers = []

for n in range(65):
    layers.append([])
    for i in range(k, 8320, 65):
        layers[n].append(i)
    k += 1

for i in range(len(layers)):
    layers[i].insert(0, layers[i][-1])
    string = str(layers[i]).strip("[]")
    newloop = "Line(" + str(i) + ") = {" + string + "};" + "\n"
    print(newloop)
```
